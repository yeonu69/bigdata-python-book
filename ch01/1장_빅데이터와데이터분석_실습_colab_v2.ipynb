#### [파이썬 실습 1-2 ★] CSV 파일 불러오기

인터넷에 있는 따릉이 CSV 파일을 판다스로 불러와 표의 크기와 열 이름을 확인하고, 첫 7일을 그리시오.

① 판다스를 불러옵니다. ② 주소에서 파일을 읽습니다. ③ 크기와 열 이름을 보고, 첫 168행을 그립니다.

CSV(comma-separated values)는 표의 칸을 쉼표로 나눠 적은 글자 파일입니다. 원본은 UCI 저장소에 있고, 여기서는 같은 파일을 올려 둔 깃허브 주소에서 바로 읽습니다.

```python
import pandas as pd      # 표를 다루는 도구 판다스를 pd라는 별명으로 불러옵니다

URL = ("https://raw.githubusercontent.com/syedsharin/"
       "Seoul-Bike-Sharing-Demand-Prediction/main/SeoulBikeData.csv")
df = pd.read_csv(URL, encoding="latin-1")     # 글자 저장 방식(인코딩)을 알려 줍니다
print(df.shape)                 # (행 수, 열 수)
print(df.columns.tolist())      # 열 이름 목록

df["Rented Bike Count"].head(168).plot()     # 앞의 168행 = 첫 7일
plt.title("첫 7일(168시간)의 시간별 대여량")
plt.xlabel("행 번호(시간)")
plt.show()
```

`import`는 도구 상자를 꺼내 오는 명령입니다. `pd.read_csv( )`는 CSV 파일을 읽어 표로 만듭니다. `encoding="latin-1"`은 글자의 저장 방식을 알려 줍니다. "이 파일은 latin-1 방식이다"라는 뜻입니다. 이것을 빼면 첫 줄의 °C 기호에서 읽기가 멈춥니다([파이썬 실습 1-8 ☆]). 표는 8,760행, 14열로 [표 1-4]와 같습니다. 데이터를 불러오면 늘 크기부터 보는 버릇을 들이십시오. 행 수가 예상과 다르면 파일이 잘렸거나 다른 파일입니다.

#### [파이썬 실습 1-3 ★] 한 행이 무엇인지 확인하기

열 이름을 짧게 바꾸고, 날짜 열을 날짜형으로 바꾸시오. 날짜마다 행이 몇 개인지 세고, 하루 대여량 365일을 그리시오.

① 열 이름 14개를 순서대로 바꿉니다. ② 날짜를 "일/월/년"으로 읽습니다. ③ 날짜마다 행 수와 대여량 합을 셉니다.

```python
df.columns = ["date", "count", "hour", "temp", "humidity", "wind", "visibility",
              "dew", "solar", "rain", "snow", "season", "holiday", "functioning"]
df["date"] = pd.to_datetime(df["date"], format="%d/%m/%Y")   # 글자 → 날짜 (일/월/년)

print(df[["date", "hour", "count", "temp", "functioning"]].head(3))
print("날짜 수:", df["date"].nunique())
print("날짜마다 행 수:", df.groupby("date").size().unique())

daily = df.groupby("date")["count"].sum()     # 날짜마다 대여량을 더하기
print("하루 대여량이 0인 날:", (daily == 0).sum(), "일")
print(daily["2018-04-09":"2018-04-15"].tolist())   # [예제 1-1]의 한 주
daily.plot()
plt.title("하루 대여량 365일")
plt.ylabel("대여량(대)")
plt.show()
```

날짜는 365개이고, 모든 날짜가 24행씩입니다. 한 행은 한 시간이라는 짐작이 확인되었습니다. 하루 합이 0인 날은 12일입니다. 마지막 줄은 [예제 1-1]의 한 주를 이 표에서 꺼낸 것입니다. `format="%d/%m/%Y"`를 빼면 12월 1일을 1월 12일로 읽을 수 있으니 조심하십시오.

> **[코드 읽기] 대괄호로 고르기와 `groupby`**
>
> `df["date"]`처럼 표 뒤에 대괄호와 열 이름을 쓰면 그 열 하나를 꺼냅니다. 대괄호를 두 겹 `[[ ]]`으로 쓰고 열 이름을 늘어놓으면 여러 열을 꺼냅니다. `df.groupby("date")["count"].sum()`은 "날짜가 같은 행끼리 묶어서, count를 더하라"는 한 줄입니다. 묶을 열, 계산할 열, 계산 방법(sum, mean)의 순서로 읽으면 됩니다.

#### [파이썬 실습 1-4 ★] 큰 수 세어 보기: 원 기록은 몇 건인가

1년 동안의 대여 건수를 모두 더해 [예제 1-3]의 어림셈과 견주시오. 계절마다 몇 %를 빌렸는지 구하고 막대그래프로 그리시오.

① count 열을 모두 더합니다. ② 어림셈과 견주고 원 기록의 크기를 어림합니다. ③ 계절로 묶어 더하고 백분율로 바꿉니다.

```python
total = df["count"].sum()
print("1년 동안 빌린 횟수:", total)
print("어림셈 17,000 × 365:", 17000 * 365)
print("한 건을 100바이트로 잡은 원 기록 크기(MB):", round(total * 100 / 1_000_000))

season = df.groupby("season")["count"].sum()
print((season / total * 100).round(1))       # 계절마다 몇 %인가
(season / 10000).plot(kind="bar", rot=0)
plt.title("계절별 대여량 합계(만 건)")
plt.show()
```

8,760칸을 더한 6,172,314건이 우리 표 뒤에 숨은 원 기록의 수입니다. 어림셈 6,205,000과 거의 같습니다. 그러니 계산이 자릿수부터 틀리지는 않았습니다. 계절의 몫은 여름 37.0%, 겨울 7.9%로 4.2절과 같습니다. 파이썬에서는 큰 수를 `1_000_000`처럼 밑줄로 끊어 써도 됩니다.

#### [파이썬 실습 1-5 ★] 운영하지 않은 날의 0 찾기

functioning 열의 값을 세고, 운영하지 않은 시간을 넣었을 때와 뺐을 때의 평균을 견주시오.

① 값마다 몇 번 나오는지 셉니다. ② 운영하지 않은 시간만 골라 날짜 수와 대여량 합을 봅니다. ③ 운영한 시간만 골라 평균을 다시 냅니다.

```python
print(df["functioning"].value_counts())
off = df[df["functioning"] == "No"]           # 운영하지 않은 시간만 고르기
print("운영하지 않은 날:", off["date"].nunique(), "일 / 그 시간의 대여량 합:", off["count"].sum())

on = df[df["functioning"] == "Yes"]           # 운영한 시간만 고르기
print("시간당 평균(모든 시간):", round(df["count"].mean(), 1))
print("시간당 평균(운영한 시간만):", round(on["count"].mean(), 1))
print("하루 평균(365일):", round(daily.mean()))
print("하루 평균(0인 날을 뺀", (daily > 0).sum(), "일):", round(daily[daily > 0].mean()))

plt.bar(["모든 시간", "운영한 시간만"], [df["count"].mean(), on["count"].mean()])
plt.title("0을 넣을 때와 뺄 때의 시간당 평균")
plt.ylabel("대여량(대)")
plt.show()
```

"No"가 295번이고, 그 시간들은 13개 날짜에 걸쳐 있습니다. 13일 가운데 10월 6일은 아침 7시간만 해당합니다. 그 시간의 대여량 합은 정확히 0입니다. 0을 넣고 빼는 것만으로 시간당 평균이 24.6대 달라집니다. `df[df["functioning"] == "No"]`는 "functioning이 No인 행만 골라 달라"는 뜻입니다. 등호 두 개 `==`는 "같은가?"를 묻습니다.

#### [파이썬 실습 1-6 ★] 첫 질문에 답하기: 시간대별 평균 대여량

운영한 시간만으로 시각(0~23시)별 평균 대여량을 구해 꺾은선 그래프로 그리시오.

① 실습 1-5의 `on`을 시각으로 묶어 평균을 냅니다. ② 24개 점을 선으로 잇습니다. ③ 가장 붐비는 시각과 가장 한가한 시각을 찾습니다.

```python
by_hour = on.groupby("hour")["count"].mean()     # 시각(0~23시)별 평균 대여량
plt.plot(by_hour.index, by_hour.values, marker="o")
plt.xticks(range(0, 24, 2))
plt.xlabel("시각(시)")
plt.ylabel("평균 대여량(대)")
plt.title("따릉이 시간대별 평균 대여량")
plt.grid(alpha=0.3)
plt.show()
print("가장 붐비는 시각:", by_hour.idxmax(), "시", round(by_hour.max()), "대")
print("가장 한가한 시각:", by_hour.idxmin(), "시", round(by_hour.min()), "대")
```

1년 치 0시끼리, 1시끼리 … 23시끼리 평균을 내니 값이 24개 나옵니다. 18시가 1,554대로 가장 높고, 새벽 4시가 137대로 가장 낮습니다. 8시에도 뾰족한 봉우리가 보입니다. [표 1-6]의 세 번째 질문에 첫 답을 얻었습니다. `idxmax( )`는 가장 큰 값이 있는 자리(여기서는 시각)를 알려 줍니다.

#### [파이썬 실습 1-7 ☆] 평일과 주말을 나눠 그리기

24시간을 모두 운영한 352일만 남겨, 평일과 주말의 시간대별 평균 대여량을 한 그래프에 그리시오.

```python
# 24시간을 모두 운영한 352일만 남기기 (2장과 같은 기준)
off_days = off["date"].unique()                       # 운영하지 않은 시간이 있는 13일
full = df[~df["date"].isin(off_days)].copy()
full["weekend"] = full["date"].dt.dayofweek >= 5      # 토요일(5)·일요일(6)이면 True
print("남은 날:", full["date"].nunique(), "/ 주말:", full[full["weekend"]]["date"].nunique())

weekday = full[~full["weekend"]].groupby("hour")["count"].mean()
weekend = full[full["weekend"]].groupby("hour")["count"].mean()
plt.plot(weekday, marker="o", label="평일")
plt.plot(weekend, marker="s", label="주말")
plt.xticks(range(0, 24, 2))
plt.xlabel("시각(시)")
plt.ylabel("평균 대여량(대)")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
print("8시 평일:", round(weekday[8]), "/ 주말:", round(weekend[8]))
print("가장 붐비는 시각 평일:", weekday.idxmax(), "시", round(weekday.max()),
      "/ 주말:", weekend.idxmax(), "시", round(weekend.max()))
```

같은 그래프가 책 03절에도 실려 있습니다. `isin( )`은 "이 목록에 들어 있나?"를 묻고, 물결표 `~`는 "아니다"를 뜻합니다. 그래서 `~df["date"].isin(off_days)`는 쉰 날이 아닌 날만 고릅니다. `dt.dayofweek`는 요일을 월요일 0부터 일요일 6까지의 숫자로 알려 줍니다. 평일 8시는 1,306대, 주말 8시는 425대로 세 배 넘게 차이 납니다. 2장도 같은 352일로 셉니다.

#### [파이썬 실습 1-8 ☆] 오류 메시지 읽기

파일을 바이트 그대로 받아 크기를 재고, 인코딩을 알려 주지 않고 읽으면 어떤 오류가 나는지 확인하시오.

#### [파이썬 실습 1-10 ☆] 서울의 네 가지 기록 열어 보기

강의 자료 파일 네 개를 코랩 왼쪽 파일 창에 올리고, 크기와 02절의 숫자를 확인하시오.

```python
power = pd.read_csv("kpx_시간별_전국_전력수요량_2021_2025.csv")
subway = pd.read_csv("서울교통공사_서울전체_일별_시간대별_승하차_2023_2024.csv")
air = pd.read_csv("에어코리아_서울_도시대기평균_시간별_2023_2026.csv")
weather = pd.read_csv("asos_서울108_시간자료_2023_2026.csv")
for name, t in [("전력", power), ("지하철", subway), ("미세먼지", air), ("날씨", weather)]:
    print(name, t.shape)

hours = [f"{h}시" for h in range(1, 25)]
top = power[hours].max(axis=1).idxmax()               # 가장 많이 쓴 날의 행 번호
print("전력 수요 최대:", power.loc[top, hours].max(), "MW /", power.loc[top, "날짜"],
      power.loc[top, hours].idxmax())
ride = subway[subway["구분"] == "승차"]
print("지하철 하루 승차 2023년 평균:", round(ride[ride["날짜"] < "2024"]["합계"].mean()))
print("초미세먼지 평균(㎍/㎥):", round(air["PM25"].mean(), 1),
      "/ 나쁨 이상(36 이상) 시간의 비율:", round((air["PM25"] >= 36).mean() * 100, 1), "%")
print("기간:", weather["일시"].min(), "~", weather["일시"].max())
print("가장 더운 시각:", weather["기온(°C)"].max(), "°C / 가장 추운 시각:", weather["기온(°C)"].min(), "°C")

daily_gwh = power[hours].sum(axis=1) / 1000          # 하루 전력량(GWh)
plt.plot(pd.to_datetime(power["날짜"]), daily_gwh)
plt.title("전국 하루 전력 수요량(GWh)")
plt.show()
```

네 파일 모두 한 행이 "하루"이거나 "한 시각"입니다. 전력 파일은 하루가 한 행이고, 1시~24시가 열로 옆에 늘어서 있습니다. 그래서 `power[hours].sum(axis=1)`로 한 행의 24칸을 더해 하루 전력량을 냅니다. 전력 파일이 1,827행인 것은 5년 치 1,826일 끝에 빈 줄이 하나 더 붙어 있기 때문입니다. 이 빈 줄은 2장 미니 프로젝트에서 다시 만납니다. 이렇게 행과 열을 가로세로로 계산하는 법은 2장에서 넘파이로 배웁니다.

## [미니 프로젝트 1] 서울 지하철 기록의 신원 조회

**질문.** 서울교통공사가 공개한 "하루 승차 인원" 파일을 처음 받았습니다. 이 숫자는 무엇을 셌을까요? 그대로 믿어도 될까요? 03절에서 따릉이 표에 던진 질문을 지하철 기록에 그대로 던져 보시오.

**데이터.** 강의 자료로 나누어 주는 「서울교통공사_서울전체_일별_시간대별_승하차_2023_2024.csv」(약 0.2MB)입니다. 원자료는 서울교통공사가 공공데이터포털에 올린 "역별 일별 시간대별 승하차인원 정보"입니다. 원자료는 역 하나, 하루, 승차 또는 하차가 한 행이고 한 해에 약 20만 행입니다. 강의 자료는 날짜와 승하차마다 1~8호선 모든 역을 더했습니다. 그래서 한 행은 "하루의 승차" 또는 "하루의 하차"이고, 1,462행 22열입니다. 열은 날짜, 구분, 시간대 19개(06시이전~23시이후), 합계입니다.


**1단계: 데이터 신원 확인**([파이썬 실습 1-2], [1-3] 응용). 파일을 읽어 크기, 기간, 날짜 수, 구분 값, 빈칸을 확인하시오.

```python
PATH = "서울교통공사_서울전체_일별_시간대별_승하차_2023_2024.csv"   # 코랩 왼쪽 파일 창에 올린 파일
sub = pd.read_csv(PATH)
print(sub.shape)
print(sub.iloc[:2, :5])
print("기간:", sub["날짜"].min(), "~", sub["날짜"].max(), "/ 날짜 수:", sub["날짜"].nunique())
print(sub["구분"].value_counts())
print("빈칸:", sub.isna().sum().sum())
```

731일 × 승차·하차 = 1,462행입니다. 2023년과 2024년이 하루도 빠짐없이 있고, 빈칸은 없습니다. 날짜마다 승차 한 행, 하차 한 행입니다.

**2단계: 이상한 달 찾기**([파이썬 실습 1-3] 응용). 달마다 13~14시 승차의 하루 평균을 그리고, 다른 달과 크게 다른 달을 찾으시오.

```python
sub["날짜"] = pd.to_datetime(sub["날짜"])
ride = sub[sub["구분"] == "승차"].copy()
ride["달"] = ride["날짜"].dt.strftime("%Y-%m")
month = ride.groupby("달")["13-14시"].mean() / 10000          # 만 명 단위
print(month.round(1).to_string())
month.plot(marker="o")
plt.title("달마다 하루 평균 13~14시 승차(만 명)")
plt.xticks(rotation=90)
plt.show()
```

다른 달은 하루 평균 20만 명 안팎입니다. 그런데 2024년 4월만 175.9만 명입니다. 한 달 사이 낮 승객이 여덟 배가 될 까닭이 있을까요? 이 값은 강의 자료를 만들다 생긴 것이 아닙니다. 공개된 원자료에 그대로 있습니다. 예를 들어 2024년 4월 2일 신림역의 13~14시 승차는 39,275명입니다. 바로 앞뒤 시간대는 2,467명과 2,260명입니다. 따릉이의 "운영하지 않은 0"처럼, 기록에 적힌 숫자가 세상과 어긋난 경우입니다.

**3단계: 기준선과 비교**([파이썬 실습 1-5] 응용). 2024년 4월을 넣을 때와 뺄 때의 하루 평균 승차를 견주시오. 이상한 달을 뺀 표로 평일과 주말의 하루 평균을 구하시오.

```python
odd = ride["달"] == "2024-04"
print("2024년 4월 하루 합계 평균(만 명):", round(ride[odd]["합계"].mean() / 10000, 1))
print("다른 달 하루 합계 평균(만 명):", round(ride[~odd]["합계"].mean() / 10000, 1))
ok = ride[~odd].copy()                                       # 이상한 달을 뺀 표
ok["주말"] = ok["날짜"].dt.dayofweek >= 5
by_day = ok.groupby("주말")["합계"].mean()
print("하루 평균 승차: 평일", round(by_day[False]), "/ 주말", round(by_day[True]))
print("주말은 평일의", round(by_day[True] / by_day[False] * 100, 1), "%")
plt.bar(["평일", "주말"], by_day.values / 10000)
plt.title("하루 평균 승차(만 명, 2024년 4월 제외)")
plt.show()
```

2024년 4월은 하루 합계도 635.2만 명으로 부풀었습니다. 다른 달은 428.8만 명입니다. 평일은 하루 평균 4,773,610번, 주말은 3,078,667번 승차했습니다. 주말은 평일의 64.5%입니다. 요일은 실습 1-7의 `dt.dayofweek`로 갈랐습니다.

**4단계: 큰 수와 어림셈**([파이썬 실습 1-4] 응용). 2023년 1년 승차 합계를 구하시오. 하루 승차가 가장 많은 날과 적은 날, 평일에 가장 붐비는 시간대도 찾으시오. 1년 합계를 어림셈으로 검사하시오.

```python
print("2023년 1년 승차 합계:", ok[ok["날짜"].dt.year == 2023]["합계"].sum())
print("하루 승차 최대:", ok.loc[ok["합계"].idxmax(), "날짜"].date(), ok["합계"].max())
print("하루 승차 최소:", ok.loc[ok["합계"].idxmin(), "날짜"].date(), ok["합계"].min())
hours_sub = sub.columns[2:21]                                # 06시이전 ~ 23시이후
peak = ok[~ok["주말"]][hours_sub].mean()
print("평일에 가장 붐비는 시간대:", peak.idxmax(), round(peak.max()))
```

2023년 1년 승차는 1,547,141,920번, 약 15.5억 번입니다. 어림셈으로 검사해 봅시다. 하루 약 420만 번 × 365일 ≈ 15.3억 번이니 자릿수가 맞습니다. 가장 적은 날 2023년 1월 22일은 설날입니다. 평일에 가장 붐비는 시간대는 18~19시입니다.

**5단계: 한계 쓰기.** 이 파일이 센 것과 세지 않은 것을 [표 1-7]처럼 표로 쓰시오. 예를 들어 승차 횟수는 사람 수가 아닙니다. 출근과 퇴근에 한 번씩 타면 2로 셉니다. 9호선과 다른 회사가 운영하는 구간도 들어 있지 않습니다.

**제출물.** 그림 2장(달마다 13~14시 승차, 평일과 주말의 하루 평균), 숫자 표 1개(2024년 4월을 넣을 때와 뺄 때, 평일과 주말), 결론 5문장 이내. 결론에는 이 파일로 답할 수 있는 질문 하나와 없는 질문 하나를 쓰시오. 2024년 4월은 5장의 미니 프로젝트에서도 다시 걸러 냅니다.

#### [파이썬 실습 2-1 ★] 변수, 리스트, for: 일주일 가계부

값 세 개를 변수에 담아 자료형을 확인하시오. 일주일 지출을 리스트에 담고, for 문으로 합계를 구하며 1만 원 이상인 지출을 골라 담으시오.

① 코랩에서 새 노트를 열고 첫 셀에서 `!pip install koreanize-matplotlib`를 실행합니다. ② 아래 코드를 다음 셀에 쓰고 Shift + Enter를 누릅니다. ③ 결과의 합계를 2.1의 손 계산과 견줍니다.

```python
import matplotlib.pyplot as plt
import koreanize_matplotlib     # 그래프의 한글이 깨지지 않게 (코랩에서 한 번 설치)

bikes = 254            # 정수(int): 2017년 12월 1일 0시의 대여량
temp = -5.2            # 실수(float): 그때 기온(°C)
season = "Winter"      # 문자열(str): 계절
print(type(bikes), type(temp), type(season), bikes > 200)
print(3 + 4, "3" + "4")

spend = [4500, 12000, 3200, 8900, 15000, 2500, 6800]   # 월~일 지출(원)
total = 0
big = []
for x in spend:            # spend에서 값을 하나씩 꺼내 x라고 부르며 되풀이
    total = total + x      # 지금까지의 합에 x를 더한다
    if x >= 10000:         # x가 1만 원 이상이면
        big.append(x)      # big 뒤에 붙인다
print("합계:", total, "/ sum():", sum(spend), "/ 평균:", round(total / len(spend), 1))
print("1만 원 이상:", big, "/ 월요일:", spend[0], "/ 일요일:", spend[-1])

plt.bar(["월", "화", "수", "목", "금", "토", "일"], spend)
plt.axhline(10000, color="red", linestyle="--")      # 1만 원 선
plt.title("일주일 지출(원)")
plt.show()
```

`type( )`은 값의 자료형을 알려 줍니다. `bikes > 200`처럼 비교하면 답이 불(True)로 나옵니다. `3 + 4`는 7이고 `"3" + "4"`는 "34"입니다. for 문으로 직접 더한 52,900원과 `sum( )`의 값이 같습니다. 1만 원 이상인 지출은 12,000원과 15,000원입니다.

> **[코드 읽기] for와 if, 그리고 들여쓰기**
>
> `for x in spend:`는 "spend에서 값을 하나씩 꺼내 x라고 부르며 아래 줄을 되풀이하라"는 뜻입니다. 되풀이할 줄은 들여쓰기(공백 네 칸)로 표시합니다. `if x >= 10000:` 아래 줄은 한 번 더 들여 씁니다. 파이썬에서 들여쓰기는 꾸밈이 아니라 문법입니다. 깊이가 "어느 명령에 속한 줄인가"를 나타냅니다. `append( )`는 리스트 뒤에 값을 하나 붙입니다.

#### [파이썬 실습 2-2 ★] 함수와 넘파이 배열: 장바구니와 학점 평균

평균을 내는 함수를 만들어 성적과 지출에 쓰시오. 가격과 수량을 넘파이 배열로 만들어 자리끼리 곱하고, 내적으로 합계를 구하시오. 리스트와 배열에 2를 곱한 결과도 견주시오.

```python
import numpy as np

def mean(values):                       # 평균을 내는 함수: 이름 붙여 다시 쓴다
    return sum(values) / len(values)

print("성적 평균:", mean([85, 92, 78]), "/ 지출 평균:", round(mean(spend), 1))

price = np.array([1500, 3200, 4800])    # 우유, 빵, 과일 한 개 값(원)
qty = np.array([4, 2, 3])               # 담은 개수
print(price * qty)                      # 자리끼리 곱하기
print((price * qty).sum(), price @ qty) # 합계 = 내적
print([1500, 3200, 4800] * 2)           # 리스트에 2를 곱하면?
print(price * 2)                        # 배열에 2를 곱하면?

score = np.array([85, 92, 78])          # 세 과목 점수
credit = np.array([3, 2, 1])            # 학점
print("학점 평균:", score @ credit, "÷", credit.sum(), "=", round(score @ credit / credit.sum(), 2))

plt.bar(["우유", "빵", "과일"], price * qty)
plt.title("품목별 금액 = 가격 × 개수 (원)")
plt.show()
```

한 번 만든 `mean( )` 함수가 성적에도 가계부에도 쓰였습니다. 1.2의 일반화입니다. for 문 없이 `price * qty` 한 줄로 세 자리의 곱이 한꺼번에 나왔습니다. `@`로 구한 내적 26,800원은 [예제 2-2]의 손 계산과 같습니다. 셋째 줄은 [표 2-2]의 함정입니다. 리스트에 2를 곱하면 리스트가 두 번 이어 붙습니다. 숫자를 계산할 때 배열을 써야 하는 까닭입니다.

> **[코드 읽기] def와 return, np.array**
>
> `def mean(values):`는 values를 입력으로 받는 함수 mean을 만듭니다. `return` 뒤의 값이 함수가 돌려주는 결과입니다. `import numpy as np`는 넘파이를 np라는 짧은 이름으로 불러옵니다. `np.array([...])`는 리스트를 넘파이 배열로 바꿉니다.

#### [파이썬 실습 2-3 ★] 따릉이를 배열로 꺼내 조건으로 고르기

따릉이 데이터를 불러와 대여량 열을 넘파이 배열로 꺼내시오. 모양과 합계를 확인하고, 0인 시간과 3,000대 이상인 시간을 세시오. 가장 붐빈 시간도 찾으시오.

```python
import pandas as pd

URL = ("https://raw.githubusercontent.com/syedsharin/"
       "Seoul-Bike-Sharing-Demand-Prediction/main/SeoulBikeData.csv")
df = pd.read_csv(URL, encoding="latin-1")
df.columns = ["date", "count", "hour", "temp", "humidity", "wind", "visibility",
              "dew", "solar", "rain", "snow", "season", "holiday", "functioning"]
count = df["count"].to_numpy()          # 시간별 대여량 8,760개
hour = df["hour"].to_numpy()            # 그 행의 시각 0~23
print(type(count), count.shape, count.dtype, count[:5])
print("합계:", count.sum(), "/ 평균:", count.mean().round(1), "/ 최댓값:", count.max())

print("0인 시간:", (count == 0).sum())  # True를 1로 세어 더한다
busy = count >= 3000                    # True/False 8,760개
print("3,000대 이상:", busy.sum(), "시간 / 그 시각:", np.unique(hour[busy]))
i = count.argmax()                      # 가장 큰 값의 자리 번호
print("가장 붐빈 시간:", df["date"][i], hour[i], "시,", count[i], "대")

plt.hist(count, bins=40)
plt.axvline(3000, color="red", linestyle="--")
plt.title("한 시간 대여량 8,760개의 분포")
plt.xlabel("한 시간 대여량(대)")
plt.ylabel("시간 수")
plt.show()
```

`count`는 넘파이 배열이고, 모양은 길이 8,760인 한 줄, 자료형은 64비트 정수입니다. 1년 합계 6,172,314번이 한 줄로 나왔습니다. `count == 0`은 숫자 하나가 아니라 True/False 8,760개를 만듭니다. True를 1로 세므로 `.sum( )`이 곧 개수, 295시간입니다. 3,000대 이상인 31시간의 시각을 모아 보니 18시 하나뿐입니다. 가장 붐빈 시간은 2018년 6월 19일 18시의 3,556대입니다.

> **[코드 읽기] 불리언 마스크 `count[count >= 3000]`**
>
> True/False 배열을 대괄호 안에 넣으면 True인 자리의 값만 골라집니다. 이것을 불리언 인덱싱이라 하고, True/False 배열을 마스크라고 부릅니다. 엑셀의 필터와 같은 일입니다. `hour[busy]`는 3,000대 이상인 시간의 시각만 고르고, `np.unique( )`는 겹치는 값을 하나로 줄입니다. `argmax( )`는 가장 큰 값이 몇 번째 자리에 있는지 알려 줍니다.

#### [파이썬 실습 2-4 ★] reshape와 axis: 8,760시간을 365일 × 24시간으로

시각이 날마다 0시부터 23시까지 차례로 있는지 확인한 뒤, 대여량을 365 × 24 행렬로 접으시오. 하루 합계를 구해 가장 많은 날과 0인 날을 찾고, 1년을 색 지도로 그리시오.

```python
print("모든 날이 0~23시 순서인가:", (hour.reshape(365, 24) == np.arange(24)).all())
M = count.reshape(365, 24)              # 24칸씩 접어 365줄로
print(M.shape)
print(M[0])                             # 첫날(2017년 12월 1일) 24시간
print(M[0, 18], M[:3, 8])               # 첫날 18시, 첫 사흘의 8시

days = df["date"].to_numpy()[::24]      # 24칸마다 하나씩: 날마다 날짜 하나
daily = M.sum(axis=1)                   # 옆으로 더하기 → 하루 합계 365개
d = daily.argmax()
print("가장 많은 날:", days[d], daily[d], "대 / 공휴일 열:", df["holiday"][d * 24])
print("공휴일로 적힌 첫 세 날:", days[df["holiday"].to_numpy()[::24] == "Holiday"][:3])
print("합계가 0인 날:", (daily == 0).sum(), "일")

plt.figure(figsize=(7, 4))
plt.imshow(M.T, aspect="auto", cmap="YlOrRd")   # 가로 365일, 세로 24시간
plt.colorbar(label="대여량(대)")
plt.xlabel("2017년 12월 1일부터 며칠째")
plt.ylabel("시각(시)")
plt.show()
```

`hour.reshape(365, 24) == np.arange(24)`는 365개 행마다 0~23과 견줍니다. `.all( )`이 True이니 모든 날이 순서대로입니다. `M[0]`은 첫날 하루 24시간입니다. 금요일인 이날도 8시(930대)와 18시(862대)에 봉우리가 있습니다. `M[:3, 8]`은 금·토·일의 8시입니다. 토요일(219대)과 일요일(85대) 아침은 훨씬 한산합니다. 가장 많은 날은 6월 13일(36,149대)이고, 공휴일 열에는 "No Holiday"로 적혀 있습니다. 합계가 0인 날은 12일입니다.

> **[코드 읽기] 슬라이싱 `[ : ]`과 `[::24]`**
>
> `M[0, 18]`처럼 쉼표로 두 번호를 주면 칸 하나입니다. `M[:3, 8]`의 `:3`은 "처음부터 3 앞까지", 곧 처음 세 행입니다. `[::24]`는 "처음부터 끝까지 24칸마다 하나씩"이라, 8,760행에서 날마다 첫 행만 고릅니다. `M.T`는 행과 열을 바꾼 행렬(전치)입니다.

#### [파이썬 실습 2-5 ★] 숨은 0 걸러 내기: 운영한 날만 다시 평균

운영 여부 열로 "24시간 모두 운영한 날" 마스크를 만드시오. 하루 평균을 0 포함과 운영한 날만으로 견주고, 6~11월의 달별 하루 평균도 견주시오.

① 운영 여부(Yes/No)를 True/False로 바꾸고 대여량과 똑같이 365 × 24로 접습니다. ② `all(axis=1)`로 "하루 24칸이 모두 True인가"를 날마다 묻습니다. ③ 달 마스크와 운영 마스크를 `&`(그리고)로 묶습니다.

```python
on = (df["functioning"] == "Yes").to_numpy().reshape(365, 24)
full = on.all(axis=1)                   # 24시간 모두 운영한 날
print("24시간 운영한 날:", full.sum(), "일 / 아닌 날:", (~full).sum(), "일")
part = ~full & (daily > 0)              # 다 운영하지 않았는데 합계는 0이 아닌 날
print("그런 날:", days[part], "/ 운영한 시간:", on[part].sum(axis=1))
print("하루 평균(365일):", daily.mean().round(), "→ (운영한 날만):", daily[full].mean().round())
print("어림 20,000 ÷ 실제:", round(20000 / daily[full].mean(), 2))

month = pd.to_datetime(days, format="%d/%m/%Y").month.to_numpy()   # 날마다 달
with0, only = [], []
for m in [6, 7, 8, 9, 10, 11]:
    pick = month == m
    with0.append(daily[pick].mean())
    only.append(daily[pick & full].mean())
    print(m, "월  0 포함:", round(with0[-1]), " 운영한 날만:", round(only[-1]), " 운영한 날:", (pick & full).sum())

x = np.arange(6)
plt.bar(x - 0.2, with0, width=0.4, label="0 포함", color="gray")
plt.bar(x + 0.2, only, width=0.4, label="운영한 날만", color="crimson")
plt.xticks(x, ["6월", "7월", "8월", "9월", "10월", "11월"])
plt.ylabel("하루 평균 대여량(대)")
plt.legend()
plt.show()
```

운영한 날은 352일이니 운영하지 않은 날은 13일입니다. 그런데 합계가 0인 날은 12일이었습니다. 하루(2018년 10월 6일)는 17시간만 운영해 합계가 0이 아니었던 것입니다. 마스크로 고른 352일의 하루 평균은 17,489대이고, 어림 2만 번은 그 1.14배입니다. 달별 숫자는 [표 2-4]와 같습니다.

> **[코드 읽기] `&`와 `~`**
>
> `&`는 두 마스크가 모두 True인 칸만 True로 남깁니다("그리고"). `~`는 True와 False를 뒤집습니다("아니다"). 그래서 `~full & (daily > 0)`는 "다 운영하지 않았는데 합계는 0이 아닌 날"입니다. 괄호를 빼면 계산 순서가 꼬이니 조건마다 괄호로 묶습니다.

#### [파이썬 실습 2-6 ★] 평일과 주말의 하루 모양

날 번호로 요일을 계산해 주말 마스크를 만드시오. 운영한 평일과 주말의 시간대별 평균을 구해 한 그래프에 그리시오.

#### [파이썬 실습 2-8 ☆] 속도 재기: 하나씩과 한꺼번에

100만 개의 제곱의 합을 리스트 반복과 넘파이로 구하고 걸린 시간을 견주시오.

```python
import time

values = list(range(1_000_000))         # 0부터 999,999까지 리스트
arr = np.arange(1_000_000)              # 같은 숫자의 넘파이 배열

def loop_sum(xs):                       # 하나씩: 제곱해서 더하기
    total = 0
    for v in xs:
        total = total + v * v
    return total

def numpy_sum(a):                       # 한꺼번에
    return (a * a).sum()

def best_time(f, data, repeat=5):       # 다섯 번 재서 가장 빠른 시간(초)
    times = []
    for _ in range(repeat):
        start = time.perf_counter()
        f(data)
        times.append(time.perf_counter() - start)
    return min(times)

print(loop_sum(values) == numpy_sum(arr))
t_loop, t_np = best_time(loop_sum, values), best_time(numpy_sum, arr)
print(f"리스트 반복 {t_loop * 1000:.1f}ms, 넘파이 {t_np * 1000:.2f}ms, 약 {t_loop / t_np:.0f}배")
```

두 방법의 답이 같고(True), 넘파이가 수십 배 빨랐습니다. 위의 ms와 배수는 이 책을 만든 컴퓨터에서 한 번 돌린 값입니다. 코랩에서 돌리면 다른 숫자가 나옵니다. `time.perf_counter( )`는 지금 시각을 아주 정밀하게 알려 줍니다. 계산 전후의 시각을 빼면 걸린 시간입니다. 한 번만 재면 우연히 느려질 수 있어 다섯 번 재서 가장 빠른 값을 씁니다. `f"…{t_loop * 1000:.1f}…"`는 중괄호 안의 값을 소수 첫째 자리까지 끼워 넣는 f-문자열입니다.
